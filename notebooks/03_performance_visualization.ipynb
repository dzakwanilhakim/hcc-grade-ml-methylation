```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Create a DataFrame with the new data
data = {
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1'],
    'Linear SVM': [87.31, 87.18, 87.31, 87.01],
    'Polynomial SVM': [88.81, 88.56, 88.81, 88.49],
    'Neural Network': [85.82, 85.54, 85.82, 85.58],
}

df = pd.DataFrame(data)

# Step 3: Plot the grouped bar chart
metrics = df['Metric']
n_metrics = len(metrics)
bar_width = 0.25

# Positions of the bars on the x-axis
r1 = np.arange(n_metrics)
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]

plt.figure(figsize=(12, 8))

# Create bars
plt.bar(r1, df['Linear SVM'], width=bar_width, label='Linear SVM', color='skyblue')
plt.bar(r2, df['Polynomial SVM'], width=bar_width, label='Polynomial SVM', color='orange')
plt.bar(r3, df['Neural Network'], width=bar_width, label='Neural Network', color='red')

# Add labels
plt.xlabel('Metrics', fontsize=16)
plt.ylabel('Evaluation Performance (%)', fontsize=16)
plt.xticks([r + bar_width for r in range(n_metrics)], metrics, fontsize=16)
plt.yticks(fontsize=16)
plt.ylim(80, 92)

# Add legend
plt.legend(fontsize=11)

# Add threshold line at y = 80
plt.axhline(y=80, color='r', linestyle='--', linewidth=1, label='Threshold = 80')

# Show the plot
plt.show()
```

```python
#import matplotlib.pyplot as plt
def feature_importance_plot(df_feats, title):
  df_excluded = df_feats.drop(columns=['sum']).head(20)
  df_excluded = df_excluded.iloc[::-1]

  # Plotting the stacked bar chart
  ax = df_excluded.plot(kind='barh', stacked=True, figsize=(3, 8))

  # Adding title and labels
  ax.set_title(title)
  ax.set_xlabel('SHAP Score')
  ax.set_ylabel('Genes')
  plt.yticks(fontsize=12)
  plt.xticks(fontsize=10)

  # Adding legend
  #ax.legend(title='Scores')

  # Show plot
  #plt.figure(figsize=(20, 10))
  plt.show()


```
